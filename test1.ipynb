```python
import json
import re

raw = "It sounds like you’ve been carrying a lot of mental weight lately— the anxiety, the rumination, and that inner pressure can make things feel especially heavy. A simple walk, even just around the block, can give your mind a gentle pause. Moving your body helps release some of that tension, brings a bit of fresh air, and often creates a small mental “reset” that makes the worries feel a little more manageable.\n\nIf you’re up for it, try stepping outside for 10‑15 minutes, maybe to a nearby park or a quiet street. Notice the sensations in your feet, the rhythm of your breathing, and the sounds around you. You don’t have to aim for a fast pace; a leisurely stroll is enough to give your thoughts some space and gently lift your mood.\n\n---\n\n[\"Tell me the benefits of going for a walk?\",\"How can I make the most of my walks?\"]"

# Split on the --- separator
parts = raw.strip().split("\n\n---\n\n")

text = parts[0].replace("\\n", "\n")  # unescape newlines if needed

# Extract the JSON array from the second part
suggestions = []
if len(parts) > 1:
    match = re.search(r'\[.*\]', parts[1], re.DOTALL)
    if match:
        suggestions = json.loads(match.group())

print(text)
print(suggestions)
# ['Tell me the benefits of going for a walk?', 'How can I make the most of my walks?']
```
